# Launch sites on an interactive Folium map

The map uses the IBM 90-row launch cohort. Site markers and 50 km reference circles are included. The course CSV has no rail, road or coastline geometry, so this notebook does not invent proximity measurements.


```python
import pandas as pd, folium
url='https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DS0321EN-SkillsNetwork/datasets/dataset_part_2.csv'
data=pd.read_csv(url)
sites=data.groupby('LaunchSite')[['Latitude','Longitude']].first().reset_index()
print(sites.to_string(index=False))
```

```python
map_view=folium.Map(location=[31.2,-94],zoom_start=4,tiles='OpenStreetMap')
for row in sites.itertuples(index=False):
 folium.Marker([row.Latitude,row.Longitude],popup=row.LaunchSite,tooltip=row.LaunchSite).add_to(map_view)
 folium.Circle([row.Latitude,row.Longitude],radius=50000,color='#0B49CB',fill=False).add_to(map_view)
map_view.save('spacex_sites.html')
print('Saved spacex_sites.html')
```
