# Data collection: SpaceX API and web sources

This notebook uses the IBM Skills Network frozen SpaceX API response. A live Wikipedia scrape was blocked in this environment, so the published IBM SQL CSV is used for the tabular web-source exercises. This limitation is explicit rather than presented as a completed live scrape.


```python
import json, urllib.request, collections
url = 'https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DS0321EN-SkillsNetwork/datasets/API_call_spacex_api.json'
with urllib.request.urlopen(url) as response:
    launches = json.load(response)
falcon9_id = '5e9d0d95eda69973a809d1ec'
falcon9 = [row for row in launches if row['rocket'] == falcon9_id]
print('All API records:', len(launches))
print('Falcon 9 API records:', len(falcon9))
print('First Falcon 9 launch:', falcon9[0]['date_utc'][:10])
```

## Web-source table

The course supplies a 101-row CSV derived from historical launch records. The following reproducible step loads that published table. The live Wikipedia page rejected automated access (HTTP 403), so this notebook does not claim a current scrape.


```python
import pandas as pd
sql_url = 'https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DS0321EN-SkillsNetwork/labs/module_2/data/Spacex.csv'
web_table = pd.read_csv(sql_url)
print('Course web-source rows:', len(web_table))
print('Columns:', ', '.join(web_table.columns))
```

The API response has 99 Falcon 9 records; the curated 90-row modelling table uses a narrower sample and should not be merged by row number.
